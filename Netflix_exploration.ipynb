```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("netflix.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python

df.dropna(subset=["date_added","rating","duration"],inplace=True)
```

```python
df.isnull().sum()
```

```python
df["director"]=df["director"].fillna("Unknown")
df["cast"]=df["cast"].fillna("Not_Available")
df["country"]=df["country"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.select_dtypes(include="int64").columns
```

```python
df.select_dtypes(include="object").columns
```

```python
df.info()
```

```python
df["date_added"]=df["date_added"].str.strip()
```

```python
df.head()
```

```python
df["date_added"]=pd.to_datetime(df["date_added"])
```

```python
df.info()
```

```python
df["year_added"]=df["date_added"].dt.year
df["month_added"]=df["date_added"].dt.month_name()
```

```python
df.head()
```

```python
# count of movies and tv shows
df["type"].value_counts()
```

```python
df["country"].value_counts()
```

```python
df["rating"].value_counts()
```

```python
# for actors analysis
cast_df=df.copy()
cast_df["cast"]=cast_df["cast"].str.split(",")
cast_df=cast_df.explode("cast")
```

```python
cast_df.head()
```

```python
# for directors analysis
director_df=df.copy()
director_df["director"]=director_df["director"].str.split(",")
director_df=director_df.explode("director")
```

```python
director_df.head(10)
```

```python
# for countries analysis
country_df=df.copy()
country_df["country"]=country_df["country"].str.split(",")
country_df=country_df.explode("country")
```

```python
country_df.head(10)
```

```python
# for genre analysis
genre_df=df.copy()
genre_df["listed_in"]=genre_df["listed_in"].str.split(",")
genre_df=genre_df.explode("listed_in")
```

```python
genre_df.head(10)
```

```python
# top 10 countries
country_df["country"]=country_df["country"].str.strip()
country_df[country_df["country"] != "Unknown"]["country"].value_counts().head(10)
```
